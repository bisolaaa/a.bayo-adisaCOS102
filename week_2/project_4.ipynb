```python
# Project IV
# Izifin Technology: Staff ATR Calculator

print("--- Izifin Technology Payroll System ---")

# Collect User Input
experience = int(input("Enter years of work experience: "))
age = int(input("Enter staff age: "))

# Selection Logic (Tax Categorization)
if experience > 25 and age >= 55:
    atr = 5600000
elif experience > 20 and age >= 45:
    atr = 4480000
elif experience > 10 and age >= 35:
    atr = 1500000
else:
    atr = 550000

# Output the result
print(f"\nThe calculated Annual Tax Revenue (ATR) is: ₦{atr:,}")
```
