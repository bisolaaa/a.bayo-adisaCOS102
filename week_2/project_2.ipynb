```python
ALGORITHM: Polynomial Root Finder

1. Start

2. Display menu:
      1. Quadratic Equation
      2. Cubic Equation
      3. Quartic Equation

3. Read user choice

4. IF choice = 1 THEN
        Prompt user for A, B, C
        Form equation: Ax² + Bx + C = 0
        Compute roots
        Display roots

5. ELSE IF choice = 2 THEN
        Prompt user for A, B, C, D
        Form equation: Ax³ + Bx² + Cx + D = 0
        Compute roots
        Display roots

6. ELSE IF choice = 3 THEN
        Prompt user for A, B, C, D, E
        Form equation: Ax⁴ + Bx³ + Cx² + Dx + E = 0
        Compute roots
        Display roots

7. ELSE
        Display “Invalid choice”

8. End
```

```python
# You have been requested to develop a program to find the roots of
# a cubic, quartic, or quadratic equation based on user choice.

import sympy as sp

print("Welcome! This program finds roots of polynomial equations.\n")
print("Choose an operation:")
print("1. Quadratic Equation")
print("2. Cubic Equation")
print("3. Quartic Equation")

choice = input("Enter your choice (1/2/3): ")

x = sp.symbols('x')

if choice == "1":
    print("\nQuadratic Equation: Ax² + Bx + C = 0")
    A = float(input("Enter A: "))
    B = float(input("Enter B: "))
    C = float(input("Enter C: "))
    roots = sp.solve(A*x**2 + B*x + C, x)
    print("Roots:", roots)

elif choice == "2":
    print("\nCubic Equation: Ax³ + Bx² + Cx + D = 0")
    A = float(input("Enter A: "))
    B = float(input("Enter B: "))
    C = float(input("Enter C: "))
    D = float(input("Enter D: "))
    roots = sp.solve(A*x**3 + B*x**2 + C*x + D, x)
    print("Roots:", roots)

elif choice == "3":
    print("\nQuartic Equation: Ax⁴ + Bx³ + Cx² + Dx + E = 0")
    A = float(input("Enter A: "))
    B = float(input("Enter B: "))
    C = float(input("Enter C: "))
    D = float(input("Enter D: "))
    E = float(input("Enter E: "))
    roots = sp.solve(A*x**4 + B*x**3 + C*x**2 + D*x + E, x)
    print("Roots:", roots)

else:
    print("Invalid choice. Please restart the program.")
```
